```python
import pandas as pd
import numpy as np
```

dropna() - NaN qiymatlarni tashlab yuborish

```python
df = pd.read_csv("https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/main/automobile_data_nan.csv")
df.head()
```

```python
df.isnull() #mavjud bo'lmagan qiymatlarga TRUE qaytaradi
```

```python
df.notnull() #mavjud bo'lmagan qiymatlarga FALSE qaytaradi
```

```python
df.isnull().sum() #harbir ustundagi mavjud bo'lmagan qiymatlar
```

```python
len(df) #dataFrame uzunligi
```

```python
df.isnull().sum()/len(df)*100
```

```python
df.dropna() #NaN qiymatga ega qatorlarni tashlab yuborish
```

```python
len(df.dropna())
```

```python
df.dropna(axis=1) #NaN qiymatli ustunlarni tashlab yuborish
```

```python
df.dropna(axis=1, thresh=48) #48 tadan kam haqiqiy qiymatga ega (ya'ni NaN'lari ko'p) ustunlar o'chiriladi
```

```python
df.dropna(subset=['price', 'horsepower'], inplace=True) #faqat price va horsepower ustunini tekshirish
df.head()
```

```python
df.shape
```

.fillna() - NaN qiymat to'ldirish

```python
df.fillna(0) #barcha NaN qiymatlarni 0 bn to'ldirish
```

```python
df.fillna({'wheel-base':100}) #'wheel-base'ustunidagi NaN qiymatlarni 100ga tengladik
```

```python
np.mean(df['wheel-base'])
```

```python
df.fillna({'wheel-base':np.mean(df['wheel-base'])}) #ustundagi NaN qiymtlarni uni o'rta qiymati bn to'ldirish
```

```python
df.head()
```

```python
df.fillna(method='ffill', inplace=True) #NaN qiymatga ega ma'lumotni unda oldingi qiymat bn tenglash
df.head()
```

```python
df.fillna(method='bfill', inplace=True) #NaN qiymatga ega ma'lumotni unda keyingi qiymat bn tenglash
df.head()
```
